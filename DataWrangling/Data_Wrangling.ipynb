#### Missing Data

```python
# replace ? with NaN (Not a Number/something that is missing)
df.replace('?', np.nan, inplace = True)
```

```python
# compute mean and median for normalized losses
avg_norm_loss = df['normalized-losses'].astype('float').mean(axis = 0)
print('Average normal losses: ', avg_norm_loss)
med_norm_loss = df['normalized-losses'].astype('float').median(axis = 0)
print('Median normal losses: ', med_norm_loss)
```

```python
# replace missing normalized losses with mean
df['normalized-losses'] = df['normalized-losses'].replace(np.nan, avg_norm_loss)
```

```python
# check if replacement went through
df.info()
```

#### Linear Regression Model: y = a + b*x

```python
# import linear regression model from scikit-learn
from sklearn.linear_model import LinearRegression
```

```python
# create the linear regression object
lm = LinearRegression()
```

```python
# set up our input parameters for Linear Regression Model
X = df_clean[['engine-size']]
Y = df_clean['price']
```

```python
# fit the linear model
lm.fit(X, Y)
```

```python
# print the intercept
print(lm.intercept_)
```

```python
# print the slope
print(lm.coef_)
```

#### Coefficient of Determination

```python
# compute the coefficient of determination
print('R^2 = ', lm.score(X, Y))
```

#### Polynomial Fit

```python
# function to plot the data
def PlotPoly(model, x, y, Name):
    x_new = np.linspace(1400, 5000, 1000)
    y_new = model(x_new)

    plt.plot(x, y, '.', x_new, y_new, '-')
    plt.title('Polynomial Fit')
    ax = plt.gca()   # retrieve a reference to the active plotting area (the axes)
    ax.set_facecolor((0.898, 0.898, 0.898))
    fig = plt.gcf()
    plt.xlabel(Name)
    plt.ylabel('Price of Cars')
    plt.show()
```

```python
# fit a quadratic
x = df_clean['curb-weight']
y = df_clean['price']
f = np.polyfit(x, y, 2) # the number 2 indicates a quadratic fit
p = np.poly1d(f) # creates a model
```

```python
# plot the function
PlotPoly(p, x, y, 'curb-weight')
```
