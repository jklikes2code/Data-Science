#### Working with Dates

```python
city = ['Lodon', 'Rio', 'Tokyo']
start_date = ['07-27-2012', '5th Aug, 2016', '23rd July, 2021']
end_date = ['12th Aug, 2012', '21-08-2016', '8th Aug, 2021']
```

```python
games = pd.DataFrame(zip(city, start_date, end_date),
                     columns = ['City', 'Start_Date', 'End_Date'])
games
```

```python
games.dtypes
```

```python
# change types
games['Start_Date'] = pd.to_datetime(games['Start_Date'], format = 'mixed') # converts Start_Date to a datetime format
games['End_Date'] = pd.to_datetime(games['End_Date'], format = 'mixed')
games['City'] = games.City.astype('string') # Reinforces City to be a str
games
```

```python
# check the data types
games.dtypes
```

```python
# find the duration of the games
games['End_Date'] - games['Start_Date']
```

```python
# add a column for the duration
games = games.assign(Duration = games['End_Date'] - games['Start_Date'])
games
```

```python
# check the data types (timedelta means difference)
games.dtypes
```
