```python
# import libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# simple plot
x = np.linspace(0, 4*np.pi, 100)
y = np.sin(x)
plt.plot(x, y, color = 'r')
plt.title('Sine Function')
plt.xlabel('X: Angle in Radians')
plt.ylabel('Sin (X)')
plt.show()
```

#### 3D Plots

```python
from mpl_toolkits.mplot3d import Axes3D

# create a figure in 3D and a 3D axes
fig = plt.figure()
ax = fig.add_subplot(111, projection = '3d')

# generate data for a 3D figure
t = np.linspace(0, 4 * np.pi, 100)
x = np.sin(t)
y = np.cos(t)
z = t

# plot 3D line
ax.plot(x, y, z)

# set the labels for the axes
ax.set_xlabel('x-axis')
ax.set_ylabel('y-axis')
ax.set_zlabel('z-axis')

# set up title
ax.set_title('3D Spiral')

# display the plot
plt.show()
```

```python
# create a meshgrid of X and Y values
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)

# define a function to plot
Z = X**2 - Y**2

# create a 3D surface plot
fig = plt.figure(figsize = (10, 8))
ax = fig.add_subplot(111, projection = '3d')

ax.plot_surface(X, Y, Z, cmap = 'viridis')

# add title and labels
ax.set_title('3D Surface Plot')
plt.show()
```

#### Animations

```python
# import libraries
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import display, Javascript, HTML
```

```python
# prepare the plot
fig, ax = plt.subplots()
xdata, ydata = [], []
line, = ax.plot([], [], 'r-') # what is the comma for? (nature's greatest mysteries
ax.set_xlim(0, 2 * np.pi)
ax.set_ylim(-1.5, 1.5)

# define the update function
def update(frame):
    xdata.append(frame)
    ydata.append(np.sin(frame))
    line.set_data(xdata, ydata)
    return line,

# construct the FuncAnimation object
ani = FuncAnimation(
    fig, 
    update,
    frames = np.linspace(0, 2*np.pi, 128),
    interval = 20, # delay between frames in milliseconds
    blit = True  # optimize the rendering
)

plt.close(fig)

# display as JavaScript
HTML(ani.to_jshtml())
```
