#### NumPy Data Types
* i - integer
* b - boolean
* u - unsigned integer (whole numbers)
* f - float
* S - string

```python
# check the type for an array
a = np.array([4, 2, 7, 5, 3, 8, 9, 1, 6])
print(a.dtype)
```

```python
# create an array of strings
b = np.array([2, 8, 5], dtype = 'S')
print(b)
print(b.dtype)
```

```python
# create an array of floats
c = np.array([1.9, 8.2, 3.7])
print(c)
print(c.dtype)
```

```python
# convert the array of floats to an array of integers
d = c.astype('i')
print(d)
print(d.dtype)
```

#### Array Shape and Reshape

```python
b = np.array([[4, 9, 2], [3, 5, 7], [8, 1, 6]])
print(b)
print(b.shape) #prints number of rows and columns
print(b.size) #prints number of elements in 2D array
```

```python
# create a 1D array
c = np.array([9, 1, 2, 8, 7, 3, 4, 6, 5])
print(c)
```

```python
# reshape the array (must use all elements in array to reshape)
d = c.reshape(3, 3)
print(d)
```

```python
# create a 2D array and flatten it
b = np.array([[4, 9, 2], [3, 5, 7], [8, 1, 6]])
f = b.reshape(1, 9)
print(f)
```

```python
g = b.ravel()
#both b.flatten() and b.reshape(-1) does same thing
print(g)
```

#### Array Iteration

```python
# create an array and iterate (go through elements one by one in order)
a = np.array([8, 4, 6])
for x in a:
    print(x)
```

```python
# create a 2D array and iterate
b = np.array([[3, 1], [5, 8]])
for x in b:
    print(x)
```

```python
for x in b:
    for y in x:
        print(y)
```

```python
# iterate through a multidimensional array using nditer
for x in np.nditer(b):
    print(x)
```

#### Array Join / Split

```python
# join two arrays
a = np.array([6, 8, 3])
b = np.array([7, 2, 5])
c = np.concatenate((a,b))
print(c)
```

```python
# join 2d arrays
c = np.array([[3, 7], [8, 1]])
d = np.array([[9, 2], [4, 6]])
e = np.concatenate((c, d), axis = 0)
print(e)
f = np.concatenate((c, d), axis = 1)
print(f)
```

```python
# stack vertically (sames as concatenate /w axis = 0)
g = np.vstack((c, d))
print(g)
```

```python
# stack horizontally (same as concatenate /w axis = 1)
h = np.hstack((c, d))
print(h)
```

```python
# split an array
j = np.array([9, 2, 7, 1, 3, 5])
k = np.array_split(j, 3) # can go over the length of array
print(k)
```

```python
print(k[2])
```

```python
# split a 2D array
m = np.array([[4, 1], [5, 7], [3, 8], [2, 9]])
print(m)
```

```python
n = np.array_split(m,2)
print(n)
```

```python
print(n[0])
```

```python
print(n[1])
```

#### Array Sort / Search

```python
# create a 1D array
a = np.array([2, 7, 1, 8, 3, 9, 5, 4, 6])
print(a)
b = np.sort(a)
print(b)
```

```python
# create a 2D array
c = np.array([[4, 9, 2], [3, 5, 7], [8, 1, 6]])
print(c)
d = np.sort(c)
print(d)
```

```python
# print the index for a given value
idx = np.where(a == 9)
print(idx)
```

```python
# find the indices of even values
idx = np.where(a%2 == 0)
print(idx)
```

#### Filter an Array

```python
# create an array
a = np.array([2, 7, 1, 8, 3, 9, 5, 4, 6])

# create the filter for the array
filter = (a%2 == 0)

# create a new array with that filter
b = a[filter]
print(b)
```
